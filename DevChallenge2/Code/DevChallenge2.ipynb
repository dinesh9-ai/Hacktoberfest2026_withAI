```python
import pandas as pd
import numpy as np
import ollama
import os
import glob
```

```python
log_dir = r'E:\Personal\Projects\Hacktoberfest2026\repo\Hacktoberfest2026_withAI\DevChallenge2\Data'
if not os.path.exists(log_dir):
    os.makedirs(log_dir)
    print(f"Created '{log_dir}' directory. Please place your run CSV files inside.")
    #exit()

all_files = glob.glob(os.path.join(log_dir, "*.csv"))
if not all_files:
    print("No CSV logs found in the 'logs' folder.")
    #exit()


all_files.sort(key=os.path.getmtime)
latest_file = all_files[-1]
history_files = all_files[:-1]
```

```python
def analyze_run(file_path):
    df = pd.read_csv(file_path)
    df['Impact_Magnitude'] = np.sqrt(df['Accel_X']**2 + df['Accel_Y']**2 + df['Accel_Z']**2)
    duration_mins = round((len(df) * 0.1) / 60, 1) 
    avg_impact = round(df['Impact_Magnitude'].mean(), 2)

    halfway = len(df) // 2
    fatigue_ratio = df['Impact_Magnitude'].iloc[halfway:].mean() / df['Impact_Magnitude'].iloc[:halfway].mean()
    
    return duration_mins, avg_impact, fatigue_ratio
```

```python
history_stats = []
if history_files:
    for f in history_files:
        duration, impact, _ = analyze_run(f)
        history_stats.append((duration, impact))
    
    avg_hist_duration = round(np.mean([x[0] for x in history_stats]), 1)
    avg_hist_impact = round(np.mean([x[1] for x in history_stats]), 2)
    history_summary = f"User has {len(history_files)} past runs on record. Historically averaging {avg_hist_duration} mins per run with an average impact force of {avg_hist_impact} m/s^2."
else:
    history_summary = "No historical data available. This is the user's first logged run."
```

```python
latest_duration, latest_impact, latest_fatigue = analyze_run(latest_file)
fatigue_status = "increased" if latest_fatigue > 1.05 else "remained stable"

print("\n🏃‍♂️ OpenStride - Interactive Analysis")
print("-" * 45)
print(f"Latest Run: {latest_duration} mins | Avg Impact: {latest_impact} m/s^2")
print(f"History: {len(history_files)} previous runs detected.")
print("-" * 45)

user_goal = input("\nWhat is your current training goal? (e.g., 'Sub 4-hour marathon'): ")
user_feedback = input("How did your body feel today? (e.g., 'Left calf tight'): ")
```

```python
system_prompt = f"""
You are an elite marathon coach running locally. You analyze offline hardware sensor data to provide training plans.

HISTORICAL BASELINE:
{history_summary}

TODAY'S RUN DATA:
- Duration: {latest_duration} minutes
- Average Impact Force: {latest_impact} m/s^2
- Form breakdown: Impact force {fatigue_status} in the second half, indicating {'leg fatigue' if fatigue_status == 'increased' else 'good endurance'}.

USER'S PERSONAL INPUT:
- Goal: {user_goal}
- Post-run feeling: {user_feedback}

First, provide a brief analysis comparing today to their history (if available). Then, give a specific plan for tomorrow. 
Keep your answers concise. You will then answer follow-up questions from the user.
"""

messages = [{'role': 'system', 'content': system_prompt}]

print("\n🤖 Analyzing kinematics and generating plan...\n")
```

```python
print ("Analyzing data locally (No internet required)...")
while True:
    # Get AI Response
    response = ollama.chat(model='llama3.1:8b', messages=messages)
    ai_reply = response['message']['content']
    
    print(f"\nCoach:\n{ai_reply}\n")
    print("-" * 45)

    messages.append({'role': 'assistant', 'content': ai_reply})
    

    user_msg = input("\nAsk a follow-up, tweak the plan, or type 'exit' to quit: ")
    
    if user_msg.lower() in ['exit', 'quit']:
        print("Saving your legs for tomorrow. Goodbye!")
        break
        
    messages.append({'role': 'user', 'content': user_msg})
```
